```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', 100)
```

## Load

```python
DATA_SOURCE = 'artifacts\\avaloka-fleet\\sub-avaloka-00\\shard.csv'


# The strings read as missing, per column, exactly as the analysis read them.
MISSING_STRINGS = {
    **dict.fromkeys(['prompt_id', 'source_benchmark', 'task_domain', 'agent_role', 'difficulty_tag', 'fallback_route'], ['']),
    **dict.fromkeys(['prompt_tokens', 'prompt_words', 'prompt_chars', 'prompt_lines', 'question_marks', 'code_fences', 'math_symbols', 'digit_count', 'uppercase_ratio', 'avg_word_length', 'golden_reference_words', 'claude_sonnet_score', 'gpt4o_score', 'llama70b_score', 'gpt4o_mini_score', 'llama8b_score', 'best_frontier_score', 'best_mid_score', 'best_economy_score', 'frontier_advantage', 'frontier_gain_over_mini', 'cost_per_1k_frontier_usd', 'cost_per_1k_economy_usd', 'frontier_cost_multiplier', 'high_compute_route'], ['', '#N/A', '#N/A N/A', '#NA', '-1.#IND', '-1.#QNAN', '-NaN', '-nan', '1.#IND', '1.#QNAN', '<NA>', 'N/A', 'NA', 'NULL', 'NaN', 'None', 'n/a', 'nan', 'null']),
}


def load() -> pd.DataFrame:
    df = pd.read_csv(
        DATA_SOURCE, sep=',', encoding='utf-8', keep_default_na=False, na_values=MISSING_STRINGS)
    return df.dropna(axis=1, how="all")  # all-empty columns, as the analysis dropped them


df = load()
print(df.shape)
df.head()
```

## Transform (Avaloka-selected)

```python
def transform(df):
    df = df.drop(columns=['code_fences'])  # constant columns
    for _c in ['prompt_id', 'source_benchmark', 'task_domain', 'agent_role', 'difficulty_tag', 'fallback_route']:
        df[_c] = df[_c].astype('string').str.strip()
    return df

df = transform(df)
print(df.shape)
```
